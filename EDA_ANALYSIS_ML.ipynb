```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("loan_data/Loan_default.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nDataset Information:")
df.info()
print("Missing values in each column:")
print(df.isnull().sum())
plt.figure(figsize=(12, 5))

missing = df.isnull().sum()

sns.barplot(x=missing.index, y=missing.values)

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=60)

plt.show()
print("Number of duplicate rows:", df.duplicated().sum())
df.describe()
df.describe().T
numerical_columns = [
    "Age",
    "Income",
    "LoanAmount",
    "CreditScore",
    "MonthsEmployed",
    "NumCreditLines",
    "InterestRate",
    "LoanTerm",
    "DTIRatio"
]

categorical_columns = [
    "Education",
    "EmploymentType",
    "MaritalStatus",
    "HasMortgage",
    "HasDependents",
    "LoanPurpose",
    "HasCoSigner"
]

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)
default_counts = df["Default"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    default_counts.values,
    labels=["No Default", "Default"],
    autopct="%1.2f%%",
    startangle=90
)

plt.title("Loan Default Distribution")

plt.show()
plt.figure(figsize=(7, 5))

sns.countplot(
    x="Default",
    data=df
)

plt.title("Loan Default Count")
plt.xlabel("Default (0 = No, 1 = Yes)")
plt.ylabel("Number of Loans")

plt.show()
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Age"],
    bins=30,
    kde=True
)

plt.title("Age Distribution of Borrowers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Income"],
    bins=30,
    kde=True
)

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.show()
plt.figure(figsize=(8, 5))

sns.histplot(
    df["CreditScore"],
    bins=30,
    kde=True
)

plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Frequency")

plt.show()
plt.figure(figsize=(8, 5))

sns.histplot(
    df["LoanAmount"],
    bins=30,
    kde=True
)

plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")

plt.show()
plt.figure(figsize=(8, 5))

sns.histplot(
    df["InterestRate"],
    bins=30,
    kde=True
)

plt.title("Interest Rate Distribution")
plt.xlabel("Interest Rate")
plt.ylabel("Frequency")

plt.show()
for column in categorical_columns:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        x=column,
        data=df
    )

    plt.title(f"{column} Distribution")
    plt.xlabel(column)
    plt.ylabel("Count")

    plt.xticks(rotation=30)

    plt.show()
    plt.figure(figsize=(8, 5))

sns.countplot(
    x="Education",
    hue="Default",
    data=df
)

plt.title("Education vs Loan Default")
plt.xlabel("Education")
plt.ylabel("Number of Loans")

plt.xticks(rotation=30)

plt.show()
plt.figure(figsize=(8, 5))

sns.countplot(
    x="EmploymentType",
    hue="Default",
    data=df
)

plt.title("Employment Type vs Loan Default")
plt.xlabel("Employment Type")
plt.ylabel("Number of Loans")

plt.show()
plt.figure(figsize=(8, 5))

sns.countplot(
    x="LoanPurpose",
    hue="Default",
    data=df
)

plt.title("Loan Purpose vs Loan Default")
plt.xlabel("Loan Purpose")
plt.ylabel("Number of Loans")

plt.show()
plt.figure(figsize=(7, 5))

sns.countplot(
    x="HasMortgage",
    hue="Default",
    data=df
)

plt.title("Mortgage Status vs Loan Default")
plt.xlabel("Has Mortgage")
plt.ylabel("Number of Loans")

plt.show()
plt.figure(figsize=(7, 5))

sns.countplot(
    x="HasDependents",
    hue="Default",
    data=df
)

plt.title("Dependents vs Loan Default")
plt.xlabel("Has Dependents")
plt.ylabel("Number of Loans")

plt.show()
loan_purpose_default = (
    df.groupby("LoanPurpose")["Default"]
    .mean() * 100
)

print(loan_purpose_default)
plt.figure(figsize=(8, 5))

loan_purpose_default.sort_values().plot(
    kind="bar"
)

plt.title("Default Rate by Loan Purpose")
plt.xlabel("Loan Purpose")
plt.ylabel("Default Rate (%)")

plt.xticks(rotation=30)

plt.show()
employment_default = (
    df.groupby("EmploymentType")["Default"]
    .mean() * 100
)

print(employment_default)

plt.figure(figsize=(8, 5))

employment_default.sort_values().plot(
    kind="bar"
)

plt.title("Default Rate by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Default Rate (%)")

plt.show()
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Default",
    y="CreditScore",
    data=df
)

plt.title("Credit Score vs Loan Default")
plt.xlabel("Default")
plt.ylabel("Credit Score")

plt.show()
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Default",
    y="Income",
    data=df
)

plt.title("Income vs Loan Default")
plt.xlabel("Default")
plt.ylabel("Income")

plt.show()

plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Default",
    y="LoanAmount",
    data=df
)

plt.title("Loan Amount vs Loan Default")
plt.xlabel("Default")
plt.ylabel("Loan Amount")

plt.show()
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Default",
    y="InterestRate",
    data=df
)

plt.title("Interest Rate vs Loan Default")
plt.xlabel("Default")
plt.ylabel("Interest Rate")

plt.show()
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Default",
    y="DTIRatio",
    data=df
)

plt.title("Debt-to-Income Ratio vs Loan Default")
plt.xlabel("Default")
plt.ylabel("DTI Ratio")

plt.show()
plt.figure(figsize=(12, 8))

correlation = df[numerical_columns + ["Default"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="CreditScore",
    y="LoanAmount",
    hue="Default",
    alpha=0.5
)

plt.title("Credit Score vs Loan Amount by Default")
plt.xlabel("Credit Score")
plt.ylabel("Loan Amount")

plt.show()
for column in numerical_columns:

    plt.figure(figsize=(8, 3))

    sns.boxplot(
        x=df[column]
    )

    plt.title(f"Outlier Analysis - {column}")
    plt.xlabel(column)

    plt.show()
    df["CreditScoreGroup"] = pd.cut(
    df["CreditScore"],
    bins=[299, 499, 599, 699, 799, 900],
    labels=[
        "300-499",
        "500-599",
        "600-699",
        "700-799",
        "800-849"
    ]
)

credit_default = df.groupby(
    "CreditScoreGroup",
    observed=True
)["Default"].mean() * 100

print(credit_default)
plt.figure(figsize=(8, 5))

credit_default.plot(kind="bar")

plt.title("Default Rate by Credit Score Group")
plt.xlabel("Credit Score Group")
plt.ylabel("Default Rate (%)")

plt.show()
eda_summary = df[numerical_columns].describe().T

eda_summary = eda_summary[
    ["mean", "std", "min", "25%", "50%", "75%", "max"]
]

eda_summary


```
